## 1. What is the average age of Netflix customers?

```python
import plotly.graph_objects as go

avg_age = df['age'].mean()

fig = go.Figure(go.Indicator(
    mode="number",
    value=avg_age,
    number={
        'valueformat': '.1f',
        'font': {'size': 64, 'color': '#ffffff', 'family': 'Arial Black'}
    },
    title={
        "text": "<b style='font-size:20px; color:#00E5FF;'>Average Customer Age</b><br><span style='font-size:14px; color:#a0a0a0;'>In Years</span>",
        "font": {'family': 'Arial'}
    },
    domain={'x': [0, 1], 'y': [0, 1]}
))

fig.update_layout(
    height=250,
    paper_bgcolor='#0d1117',  # خلفية داكنة فخمة (GitHub Dark style)
    plot_bgcolor='#0d1117',
    margin=dict(t=40, b=40, l=40, r=40)
)

fig.show()
```

## 2. What is the average number of watch hours?

*   List item
*   List item



```python
import plotly.graph_objects as go

avg_watch = df['watch_hours'].mean()

fig = go.Figure(go.Indicator(
    mode="number+gauge",
    value=avg_watch,
    number={
        'valueformat': '.2f',
        'suffix': ' hrs',
        'font': {'size': 50}
    },
    title={
        "text": "Average Watch Hours - Across all users",
        "font": {'family': 'Arial', 'size': 18, 'color': 'red'}
    },
    gauge={
        'axis': {'range': [0, 30]},
        'bar': {'color': "red"}
    }
))

fig.update_layout(
    height=300,
    template='plotly_dark',
    margin=dict(t=50, b=30, l=40, r=40)
)

fig.show()
```

## 3. Who are the top 10 customers with the highest watch hours?

```python
import plotly.express as px

# 1. استخراج أعلى 10 واختصار الـ Customer ID الطويل
top_10 = df.nlargest(10, "watch_hours").copy()
top_10['Customer_Short'] = top_10['customer_id'].astype(str).str[:8] # هناخد أول 8 حروف فقط

# 2. رسم الشارت
fig = px.bar(
    top_10,
    x='watch_hours',
    y='Customer_Short', # استخدمنا العمود المختصر
    orientation='h',
    color='watch_hours',
    color_continuous_scale='Reds', # تدرج جاهز وبسيط
    title='Top 10 Customers by Watch Hours',
    text_auto='.1f' # إظهار الرقم على العمود بوضوح
)

# 3. التنسيق (توسيط العنوان، إخفاء عنوان المحور، وضبط الهوامش)
fig.update_layout(
    yaxis={'categoryorder': 'total ascending'},
    title_x=0.5,               # توسيط العنوان
    title_font_size=20,        # تكبير خط العنوان
    yaxis_title="",            # إخفاء كلمة Customer ID لمنع الزحمة
    xaxis_title="Watch Hours",
    template='plotly_white',   # ثيم فاتح ونظيف
    height=500,
    margin=dict(l=80)          # تظبيط الهامش الشمال
)

fig.show()
```

## 4. Which subscription type has the most customers?

```python
subscription_count = df["subscription_type"].value_counts().reset_index()
subscription_count.columns = ['subscription_type', 'count']
fig = px.pie(subscription_count, values='count', names='subscription_type',
             hole=0.4, color_discrete_sequence=[netflix_black, netflix_red, '#b20710'],
             title="Customers by Subscription Type")
fig.update_traces(textposition='inside', textinfo='percent+label',
                  marker=dict(line=dict(color='#ffffff', width=2)))
fig.update_layout(annotations=[dict(text='Subs', x=0.5, y=0.5, font_size=20, showarrow=False)])
fig.show()
```

## 6. Which device is used by the most customers?

## 26. What is the average watch hours for active and churned customers?

```python
watch_by_churn = df.groupby("churned")['watch_hours'].mean().reset_index()
watch_by_churn['Status'] = watch_by_churn['churned'].map({0: 'Active (0)', 1: 'Churned (1)'})
fig = px.bar(watch_by_churn, x='Status', y='watch_hours', color='Status',
             color_discrete_sequence=[netflix_black, netflix_red], text_auto='.2f',
             title="Average Watch Hours: Active vs Churned Customers")
fig.update_traces(textposition='outside')
fig.update_layout(plot_bgcolor='rgba(0,0,0,0)')
fig.show()
```

## 27. What is the average number of last login days for active and churned customers?

## 28. How many customers are active and how many have churned?

## 29. Which region-subscription combinations have the highest average watch hours?
